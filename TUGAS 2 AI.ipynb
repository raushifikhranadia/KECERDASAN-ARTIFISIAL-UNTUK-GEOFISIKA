# Prompt 1 — Inversi dan Interpretasi Data Geolistrik Resistivitas 2D

# R — Role
Anda berperan sebagai ahli geolistrik dan inversi geofisika yang membimbing mahasiswa S1 Geofisika dalam mengolah serta menginterpretasi data geolistrik resistivitas 2D.

# T — Task
Lakukan analisis dan bimbing langkah-langkah pemodelan inversi dari data resistivitas semu (apparent resistivity) 2D untuk memperoleh penampang resistivitas sebenarnya di bawah permukaan.

# C — Context
Survei geolistrik resistivitas 2D digunakan untuk memetakan struktur bawah permukaan, seperti akuifer air tanah, zona resapan, atau batas batuan dasar. Analisis ini dilakukan dalam mata kuliah Eksplorasi Geolistrik untuk memahami hubungan antara nilai resistivitas semu, proses inversi, dan penampang geologi/hidrogeologi aktual.

# I — Input
Gunakan data yang diberikan berupa data lintasan pengukuran geolistrik (konfigurasi elektroda, spasi elektroda, posisi X, nilai resistivitas semu, dan nilai kesalahan/error pengukuran).

# K — Constraints
- Jangan mengubah atau merekayasa data pengukuran asli.
- Jelaskan pemilihan parameter damping factor atau regularisasi pada proses inversi.
- Tunjukkan nilai RMS error dan jelaskan batas toleransi yang layak secara fisis.
- Jangan menentukan litologi/jenis batuan secara absolut tanpa mengaitkannya dengan kisaran fisis nilai resistivitas material bumi.

# O — Output
Hasil harus mencakup:
1. Deskripsi data lintasan, konfigurasi elektroda yang digunakan, dan rentang nilai resistivitas semu.
2. Penjelasan rinci langkah-langkah pemodelan inversi beserta parameter yang dipilih.
3. Evaluasi kriteria kecocokan (fit) antara data observasi dan kalkulasi (RMS Error).
4. Penampang hasil inversi resistivitas 2D beserta pembagian zonasi resistivitas (tinggi, sedang, rendah).
5. Interpretasi geologi/hidrogeologi secara hati-hati dalam bentuk paragraf deskriptif.
6. Keterbatasan hasil dan kesimpulan singkat.

# Q — Quality Criteria
Pemodelan matematis stabil, batasan fisis resistivitas realistis, pembahasan istilah inversi dapat dipahami oleh mahasiswa S1, serta pembahasan sesuai dengan batas kemampuan interpretasi 2D.

# V — Verification
Periksa apakah penampang hasil inversi bebas dari artefak numerik yang ekstrem, nilai RMS error berada pada kisaran wajar (<5%), dan zonasi resistivitas selaras dengan kondisi geologi regional area studi.

# Prompt 2 — Analisis Nilai-b Gutenberg-Richter dan Kelengkapan Katalog Gempa

# R — Role
Anda berperan sebagai ahli seismologi kuantitatif yang membimbing mahasiswa S1 Geofisika dalam menganalisis statistik seismisitas dan kelengkapan katalog gempa.

# T — Task
Analisis katalog gempa untuk menentukan magnitudo kelengkapan ($M_c$) dan menghitung nilai-$b$ (b-value) serta nilai-$a$ (a-value) menggunakan relasi Gutenberg-Richter pada wilayah studi.

# C — Context
Relasi Gutenberg-Richter ($\log_{10} N = a - bM$) digunakan untuk menggambarkan distribusi magnitudo terhadap frekuensi kejadian gempa. Nilai-$b$ mencerminkan rasio antara gempa kecil dan gempa besar yang berkaitan dengan kondisi tegangan tektonik regional. Analisis ini dilakukan dalam mata kuliah Seismologi untuk memahami sifat statistik katalog gempa sebelum digunakan dalam pemodelan kebencanaan.

# I — Input
Gunakan katalog gempa yang diberikan, mencakup waktu kejadian, koordinat episenter, kedalaman, magnitudo ($M$), dan jenis magnitudo. Gunakan hanya data yang tersedia pada katalog.

# K — Constraints
- Jangan mengubah nilai asli data atau membuang data tanpa alasan kriteria kelengkapan.
- Tentukan magnitudo kelengkapan ($M_c$) secara objektif (misalnya menggunakan metode Maximum Curvature atau Maximum Likelihood Estimation) dan jelaskan metodenya.
- Abaikan data dengan $M < M_c$ saat menghitung nilai-$b$ agar hasil tidak terdistorsi oleh ketidaklengkapan pencatatan.
- Jangan menyamakan jenis magnitudo yang berbeda (misal: $M_w$, $M_L$, $m_b$) tanpa melakukan konversi/penjelasan terlebih dahulu.
- Jangan membuat klaim atau prediksi waktu terjadinya gempa di masa depan.

# O — Output
Hasil harus mencakup:
1. Deskripsi singkat katalog (periode waktu, cakupan wilayah, dan total kejadian awal).
2. Penentuan nilai $M_c$ beserta penjelasan cara/alasan pemilihannya.
3. Grafik/Tabel kurva Gutenberg-Richter (kumulatif dan non-kumulatif) serta jumlah data yang digunakan setelah pemotongan $M_c$.
4. Nilai-$a$ dan nilai-$b$ beserta standar eror perhitungannya.
5. Interpretasi kondisi tegangan tektonik regional berdasarkan nilai-$b$ yang diperoleh dalam bentuk paragraf.
6. Keterbatasan analisis statistik dan kesimpulan singkat.

# Q — Quality Criteria
Perhitungan parameter statistik benar secara matematis, grafik/tabel relevan, penggunaan istilah seismologi tepat, serta penjelasan mudah dipahami oleh mahasiswa S1 sebagai bahan laporan praktikum.

# V — Verification
Periksa apakah nilai-$b$ berada dalam rentang fisis yang umum (sekitar $0.7 - 1.3$), apakah data yang dipakai hitung nilai-$b$ sudah terbebas dari gempa di bawah $M_c$, dan apakah kesimpulan didukung penuh oleh hasil perhitungan numerik.

# Prompt 3 — Analisis Anisotropi Spasial dan Kriging 3D

# R — Role
Anda berperan sebagai ahli geostatistika minyak dan gas / tambang yang membimbing mahasiswa S1 Geofisika dalam memodelkan kontinuitas spasial 3D dan anisotropi data.

# T — Task
Analisis keberadaan anisotropi (perbedaan variasi berdasarkan arah) pada dataset 3D, pasang (fit) model variogram terarah (directional variogram), dan terapkan metode *Ordinary Kriging* 3D untuk memetakan sebaran nilai properti geofisika.

# C — Context
Properti fisik batuan di bawah permukaan (seperti porositas atau permeabilitas) jarang bersifat isotropik; nilainya sering kali memiliki jarak korelasi yang lebih panjang searah perlapisan dibanding arah vertikal. Analisis ini dilakukan dalam mata kuliah Geostatistika untuk memahami pentingnya memperhitungkan anisotropi sebelum membangun model reservoir 3D.

# I — Input
Gunakan dataset 3D yang diberikan berupa koordinat ($X$, $Y$, $Z$ atau kedalaman) beserta nilai variabel geofisika (seperti nilai porositas log sumur atau nilai impedansi akustik 3D). Perhatikan satuan koordinat horizontal dan vertikal.

# K — Constraints
- Jangan mengubah nilai asli data tanpa alur pre-processing yang transparan.
- Pastikan satuan jarak vertikal dan horizontal telah diselaraskan atau disesuaikan (*vertical exaggeration*) sebelum menghitung matriks jarak 3D.
- Jelaskan pemilihan parameter variogram (panjang *lag*, *lag tolerance*, dan *bandwidth*) untuk setiap arah pencarian (azimuth dan dip).
- Tunjukkan bukti keberadaan anisotropi (misalnya perbedaan nilai *range* pada arah utama vs arah tegak lurusnya).
- Jangan mengestrapolasi hasil Kriging jauh melebihi batas kedalaman atau area jangkauan data sumur/pengukuran.

# O — Output
Hasil harus mencakup:
1. Deskripsi dataset 3D dan statistik deskriptif dasar (mean, varians, skewness).
2. Analisis variogram terarah (eksperimental) pada beberapa arah azimuth dan vertikal.
3. Penentuan parameter anisotropi (rasio *anisotropy range*, arah sumur utama/major, minor, dan vertikal).
4. Pemilihan model variogram teoritis (misal: Spherical, Exponential, atau Gaussian) beserta kecocokannya.
5. Peta/penampang hasil estimasi Kriging 3D beserta peta *kriging variance* (ketidakpastian).
6. Evaluasi hasil menggunakan validasi silang (*cross-validation* / *leave-one-out*).
7. Kesimpulan dan implikasi anisotropi terhadap model geofisika.

# Q — Quality Criteria
Konsep anisotropi dijelaskan secara intuitif dan matematis yang tepat, parameter spatial dinyatakan konsisten dengan satuan meter, peta ketidakpastian dianalisis secara jujur, dan output siap dipakai sebagai bab analisis dalam laporan geostatistika.

# V — Verification
Periksa apakah nilai *range* pada arah mayor lebih besar daripada arah minor/vertikal, apakah nilai *kriging variance* terkecil berada di sekitar titik sampel data, dan apakah hasil *cross-validation* menunjukkan korelasi yang kuat antara nilai observasi dan estimasi.

# Prompt 4 — Model Subsiden Litosfer Samudra dan Topografi Batimetri

# R — Role
Anda berperan sebagai ahli geodinamika dan geofisika komputasi yang membimbing mahasiswa S1 Geofisika dalam memodelkan hubungan antara pendinginan litosfer dan kedalaman dasar laut.

# T — Task
Jelaskan dan analisis model kuantitatif subsiden termal litosfer 1D, lalu hitung perubahan kedalaman dasar laut (batimetri) sebagai fungsi dari umur lempeng samudra menggunakan pendekatan model setengah ruang (*Half-Space Cooling Model*).

# C — Context
Seiring bergeraknya lempeng samudra menjauhi punggung tengah samudra (Mid-Ocean Ridge), litosfer mengalami pendinginan, kontraksi termal, dan peningkatnya densitas yang menyebabkan dasar laut mengalami penyesuaian isostasi (subsiden). Analisis ini dilakukan dalam mata kuliah Geodinamika dan Tektonofisika untuk memahami evolusi topografi dasar laut dan struktur termal bumi.

# I — Input
Gunakan parameter termal dan fisik yang diberikan, seperti koefisien ekspansi termal ($\alpha$), densitas mantel ($\rho_m$), densitas air laut ($\rho_w$), suhu mantel ($T_m$), suhu permukaan ($T_0$), difusivitas termal ($\kappa$), dan rentang umur lempeng ($t$ dari 0 hingga 80 juta tahun).

# K — Constraints
- Jangan mengubah parameter termal dasar tanpa memberi catatan transparan.
- Turunkan/jelaskan hubungan proporsionalitas kedalaman dasar laut terhadap akar umur lempeng ($d(t) \propto \sqrt{t}$) dari prinsip kesetimbangan isostasi Isostasi Pratt/Airy.
- Pastikan semua satuan fisik (seperti $m^2/s$, $kg/m^3$, $1/K$, dan $s$) dikonversi secara konsisten dari umur dalam juta tahun (Ma).
- Batasi analisis pada model *Half-Space Cooling* (umur < 80 Ma) dan jelaskan mengapa model ini mulai tidak akurat untuk lempeng yang sangat tua (> 80 Ma).
- Jangan menyimpulkan kondisi geofisika lokal yang kompleks (seperti keberadaan *plume* atau *hotspot*) dari model ideal ini.

# O — Output
Hasil harus mencakup:
1. Penjelasan konsep fisik subsiden termal dan kesetimbangan isostasi.
2. Formulasi persamaan kedalaman dasar laut $d(t)$ sebagai fungsi dari umur lempeng $t$.
3. Tabel/Grafik kalkulasi profil kedalaman dasar laut untuk beberapa variasi umur lempeng (misal: 10, 30, 50, dan 70 Ma).
4. Perbandingan hasil kalkulasi analitik dengan data observasi batimetri rata-rata samudra.
5. Paragraf analisis interpretasi geodinamika mengenai laju pembentukan litosfer dan penurunan dasar laut.
6. Keterbatasan model *Half-Space Cooling* dibanding *Plate Model* dan kesimpulan singkat.

# Q — Quality Criteria
Penjelasan mekanika kontinuum dan isostasi tepat, konversi satuan waktu dari tahun ke detik presisi, angka kedalaman dalam satuan meter dinyatakan dengan jelas, serta keterbatasan model fisik dibahas secara kritis.

# V — Verification
Periksa apakah kedalaman dasar laut bertambah seiring bertambahnya umur lempeng, apakah nilai kedalaman mendekati nilai empiris umum (misal: ~2.5 km di punggung samudra dan bertambah seproporsional $\sqrt{t}$), serta pastikan seluruh variabel matematis terdefinisi dengan baik beserta satuannya.

# Prompt 5 — Inversi Fasies Seismik 3D Menggunakan Machine Learning

# R — Role
Anda berperan sebagai ahli AI Komputasi Geofisika yang membimbing mahasiswa S1 Geofisika dalam menerapkan algoritma Supervised Machine Learning untuk pemetaan fasies seismik 3D.

# T — Task
Jelaskan, bangun, dan evaluasi pendekatan Machine Learning (misal: Random Forest atau XGBoost) untuk mengklasifikasikan fasies/litologi seismik dari atribut seismik 3D, serta bandingkan performanya dengan metode pembanding sederhana (seperti Decision Tree dangkal atau Logistic Regression).

# C — Context
Atribut seismik (seperti *envelope*, *instantaneous frequency*, dan *acoustic impedance*) menyimpan informasi karakteristik reservoar bawah permukaan. Analisis ini dilakukan dalam mata kuliah AI untuk Geofisika untuk memetakan penyebaran fasies batuan secara otomatis pada kubus seismik 3D berdasarkan data sumur yang telah diikat (*well-tied data*).

# I — Input
Gunakan dataset yang diberikan berupa tabel fitur atribut seismik hasil ektraksi pada titik-titik sumur, mencakup koordinat ($X, Y, Z$), nilai-nilai atribut seismik, dan label kelas fasies/litologi hasil interpretasi log sumur.

# K — Constraints
- Jangan mengubah atau merekayasa nilai atribut asli tanpa penjelasan pra-pemrosesan (seperti standar deviasi/scaling).
- Pemisahan data latih (*training set*) dan data uji (*test set*) harus berbasis sumur (*blind-well test*), bukan pembagian acak per baris (*random row splitting*), untuk mencegah kebocoran spasial (*spatial data leakage*).
- Dilarang menggunakan koordinat ($X, Y, Z$) atau nama sumur sebagai fitur masukan model.
- Jelaskan prosedur penanganan data pencilan (*outlier*) atau atribut yang saling berkolerasi tinggi (*multicollinearity*).
- Hindari mengklaim bahwa model ini dapat diterapkan secara universal pada lapangan seismik lain tanpa pelatihan ulang (*retraining*).

# O — Output
Hasil harus mencakup:
1. Deskripsi dataset, jumlah sampel per sumur, dan distribusi proporsi kelas fasies.
2. Metodologi pra-pemrosesan data dan justifikasi pemilihan atribut seismik utama (*feature selection/importance*).
3. Penjelasan arsitektur model pembanding sederhana dan model Machine Learning utama yang dipilih.
4. Hasil evaluasi komparatif menggunakan *Confusion Matrix*, *F1-Score* per kelas fasies, dan *Balanced Accuracy*.
5. Analisis kesalahan (*error analysis*): fasies mana yang sering saling tumpang tindih (*overlap*) dan alasannya secara kemiripan respon elastis/seismik.
6. Interpretasi hasil pemetaan pada penampang seismik dalam bentuk paragraf serta kesimpulan singkat.

# Q — Quality Criteria
Evaluasi terbebas dari kebocoran data spasial, metrik performa disajikan per kelas litologi/fasies, penjelasan ilmiah mudah dipahami oleh mahasiswa S1, dan hasil dapat disajikan langsung dalam laporan praktikum atau tugas akhir.

# V — Verification
Periksa apakah sumur uji (*blind well*) benar-benar tidak pernah terlihat oleh model saat proses pelatihan, apakah nilai *F1-Score* model utama lebih unggul dibanding model pembanding, dan apakah klasifikasi fasies konsisten secara spasial pada penampang seismik.
